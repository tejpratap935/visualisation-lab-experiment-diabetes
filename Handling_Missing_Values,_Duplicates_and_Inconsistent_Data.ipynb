```python
import pandas as pd
import numpy as np

students = pd.DataFrame({
    "Student_ID":[101, 102, 102, 103, 104, 105],
    "Name":["Tej", "Ujjawal","Ujjawal ", "Sakshi", "Yogesh", None],
    "Dept":["CSE", "cse", "cse", "Data Science", "DATA SCIENCE", "ECE"],
    "Marks":[85,72, 72, np.nan, 88, 91],
    "Attendance": [90, 82, 82, 75, np.nan, 88],
    "Exam_Date":["2026-09-01", "01/09/2026", "01/09/2026",
      "2026-09-02","not known", "20266-09-03"]
})
```

# **Removing Missing Values with dropna()**

```python
df_clean = students.dropna()
print(df_clean)
```

```python
df_clean = students.dropna(axis=1)
print(df_clean)
```

```python
clean = students.copy()
clean["Marks"] = clean["Marks"].fillna(clean["Marks"].mean())
clean["Attendance"] = clean["Attendance"].fillna(clean["Attendance"].mean())
clean["Name"] = clean["Name"].fillna("Unknown")
print(clean[["Name","Marks","Attendance"]])
```

```python
clean = students.copy()
clean["Marks"] = clean["Marks"].fillna(clean["Marks"].median())
clean["Attendance"] = clean["Attendance"].fillna(clean["Attendance"].median())
clean["Name"] = clean["Name"].fillna("Unknown")
print(clean[["Name","Marks","Attendance"]])
```

# **Remove exact dulicate rows**

```python
print("Exact duplicates:", clean.duplicated().sum())
no_exact_duplicates = clean.drop_duplicates()
print("Rows before:", len(clean))
print("Rows after:", len(no_exact_duplicates))
print("After removing exact duplicates:", no_exact_duplicates.shape[0])
```

# **Remove duplicates using an ID**

```python
unique_students =no_exact_duplicates.drop_duplicates(
    subset=["Student_ID"], keep="last"
)
print("Rows before:", len(no_exact_duplicates))
print("Rows after:", len(unique_students))
print("After removing duplicates:", unique_students.shape[0])
print(unique_students[["Student_ID","Name","Dept"]])
```

# **Fix inconsistent text**

```python
unique_students = unique_students.copy()
unique_students["Name"] = unique_students["Name"].str.strip().str.title()
print(unique_students["Name"])
```

```python
unique_students["Dept"] = unique_students["Dept"].str.strip().str.lower()
print(unique_students["Dept"])
```

```python
dept_map ={
    "cse":"CSE",
    "ece":"ECE",
    "data science":"Data Science"
}
unique_students["Dept"] = unique_students["Dept"].map(dept_map).fillna("Needs review")
print(unique_students[["Name","Dept"]])
```

# **Convert dates and chec ranges**

```python
unique_students["Exam_Date"] = pd.to_datetime(unique_students["Exam_Date"], errors="coerce", dayfirst=True)
print(unique_students["Exam_Date"])
```

```python
print("Missing or invalid dates:", unique_students["Exam_Date"].isnull().sum())
```

```python
bad_marks = unique_students[~unique_students["Marks"].between(0,100)]
print("Marks outside 0-100")
print(bad_marks)
```

# **Simple Statistical Imputation**

# **Mean imputation**

```python
from sklearn.impute import SimpleImputer
df_mean = students.copy()
mean_imputer = SimpleImputer(strategy="mean")
df_mean[["Marks","Attendance"]] = mean_imputer.fit_transform(df_mean[["Marks","Attendance"]])
print(df_mean[["Marks","Attendance"]])
```

```python
from sklearn.impute import SimpleImputer
df_median = students.copy()
median_imputer = SimpleImputer(strategy="median")
df_median[["Marks","Attendance"]] = median_imputer.fit_transform(df_median[["Marks","Attendance"]])
print(df_median[["Marks","Attendance"]])
```

```python
from sklearn.impute import SimpleImputer
df_mode = students.copy()
mode_imputer = SimpleImputer(missing_values=None, strategy="most_frequent")
df_mode[["Name"]] = mode_imputer.fit_transform(df_mode[["Name"]])
print(df_mode)
```

```python
from sklearn.impute import SimpleImputer
df_const = students.copy()
constant_imputer = SimpleImputer(strategy="constant", fill_value=0)
df_const[["Marks", "Attendance"]] = constant_imputer.fit_transform(df_const[["Marks", "Attendance"]])
print(df_const)
```

KNN Imputation

```python
from sklearn.impute import KNNImputer
df_knn = students.copy()
knn_imputer = KNNImputer(n_neighbors=2)
df_knn[["Marks", "Attendance"]] = knn_imputer.fit_transform(df_knn[["Marks", "Attendance"]])
print(df_knn)
```
