#Reshaping array

```python
data= np.arange(1,13)
a=data.reshape(3,4)
b=data.reshape(4,3)
c=data.reshape(2,-1)
print(f'1D:{data}')
print(f'3x4:\n{a}')
print(f'4x3:\n{b}')
print(f'2xinfer:\n{c}')

print(f'A shape:\n{a.shape}')
print(f'B shape:\n{b.shape}')
print(f'C shape:\n{c.shape}')
```

#flatten() and ravel()

```python
data = np.array([[[1,2,3],[4,5,6]]])
print("Data shape:",data.ndim)
flat1= data.flatten()
print("Data shape:",flat1.ndim)
flat2= data.ravel()
print("Data shape:",flat2.ndim)
```

#Broadcasting

```python
arr= np.array([1,2,3])
res= arr+1
print(res)
```

```python
a=np.array([2,4,6])
b=np.array([[1,3,5],[7,9,11]])
res=a+b
print(res)
```

broadcasting in conditional operation

```python
a=np.array([12,24,35,45,60,72])
b=np.array(["Adult","Minor"])
res=np.where(a>18,b[0],b[1])
print(res)
```
