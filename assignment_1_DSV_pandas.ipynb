##A. Loading and Exploring the Dataset

```python
import pandas as pd
# 1. Load the students.csv file into a DataFrame named df
df = pd.read_csv("students (1).csv")

# 2. Display the first 5 rows
print("First 5 rows:")
print(df.head())

# 3. Display the last 10 rows
print("\nLast 10 rows:")
print(df.tail(10))

# 4. Find the number of rows and columns
print("\nNumber of rows and columns:")
print(df.shape)

# 5. Display all column names
print("\nColumn names:")
print(df.columns)

# 6. Display the data type of every column
print("\nData types:")
print(df.dtypes)

# 7. Display a concise summary of the dataset
print("\nDataset information:")
df.info()

# 8. Generate descriptive statistics for all numerical columns
print("\nDescriptive statistics:")
print(df.describe())

# 9. Check whether the dataset contains any missing values
print("\nMissing values in each column:")
print(df.isnull().sum())

print("\nDoes the dataset contain any missing values?")
print(df.isnull().values.any())

# 10. Find the number of unique values in each column
print("\nNumber of unique values in each column:")
print(df.nunique())
```

##B. Selecting and Accessing Data

```python
# 11. Select and display only student_name, gender, and semester
print("Student name, gender, and semester:")
print(df[["student_name", "gender", "semester"]])


# 12. Display all students who are in Semester 3
print("\nStudents in Semester 3:")
print(df[df["semester"] == 3])


# 13. Display all female students
print("\nFemale students:")
print(df[df["gender"] == "Female"])


# 14. Find students whose previous_cgpa is greater than 8.0
print("\nStudents with previous CGPA greater than 8.0:")
print(df[df["previous_cgpa"] > 8.0])


# 15. Display students whose attendance_percent is below 75%
print("\nStudents with attendance below 75%:")
print(df[df["attendance_percent"] < 75])


# 16. Find students who study more than 4 hours per day
print("\nStudents studying more than 4 hours per day:")
print(df[df["study_hours_per_day"] > 4])


# 17. Display students with python_skill_level equal to 'Advanced'
print("\nStudents with Advanced Python skill level:")
print(df[df["python_skill_level"] == "Advanced"])


# 18. Display all students who obtained grade 'A'
print("\nStudents with grade A:")
print(df[df["grade"] == "A"])


# 19. Find students whose risk_level is 'High'
print("\nStudents with High risk level:")
print(df[df["risk_level"] == "High"])


# 20. Display student_name and predicted_score for students
#     with predicted_score greater than 80
print("\nStudents with predicted score greater than 80:")
print(df.loc[df["predicted_score"] > 80, ["student_name", "predicted_score"]])
```

##C. Filtering with Multiple Conditions

```python
# 21. Students in Semester 4 with attendance_percent greater than 80%
print("21. Semester 4 students with attendance above 80%:")
print(df[(df["semester"] == 4) & (df["attendance_percent"] > 80)])


# 22. previous_cgpa > 8.0 AND predicted_score > 75
print("\n22. Students with CGPA > 8.0 and predicted score > 75:")
print(df[(df["previous_cgpa"] > 8.0) & (df["predicted_score"] > 75)])


# 23. Students in Semester 2 OR Semester 4
print("\n23. Students in Semester 2 or Semester 4:")
print(df[df["semester"].isin([2, 4])])


# 24. Female students with Advanced Python skill
print("\n24. Female students with Advanced Python skill:")
print(df[(df["gender"] == "Female") &
         (df["python_skill_level"] == "Advanced")])


# 25. Students whose study hours are between 2 and 5 hours
print("\n25. Students studying between 2 and 5 hours per day:")
print(df[df["study_hours_per_day"].between(2, 5)])


# 26. Students whose grade is A or B
print("\n26. Students with grade A or B:")
print(df[df["grade"].isin(["A", "B"])])


# 27. Students whose risk level is Medium or High
print("\n27. Students with Medium or High risk level:")
print(df[df["risk_level"].isin(["Medium", "High"])])


# 28. Students with mathematics skill Good or Excellent
print("\n28. Students with Good or Excellent mathematics skill:")
print(df[df["mathematics_skill"].isin(["Good", "Excellent"])])


# 29. Students who completed at least 2 projects
#     AND have engagement_score above 70
print("\n29. Students with at least 2 projects and engagement score above 70:")
print(df[(df["projects_completed"] >= 2) &
         (df["engagement_score"] > 70)])


# 30. Students with attendance > 85%, study hours > 3,
#     AND grade A
print("\n30. Students meeting all three conditions:")
print(df[(df["attendance_percent"] > 85) &
         (df["study_hours_per_day"] > 3) &
         (df["grade"] == "A")])
```

## D. Sorting and Ranking

```python

# 31. Sort the DataFrame by predicted_score in descending order
print("31. Students sorted by predicted_score (descending):")
print(df.sort_values(by="predicted_score", ascending=False))


# 32. Display the top 10 students according to predicted_score
print("\n32. Top 10 students by predicted_score:")
print(df.nlargest(10, "predicted_score"))


# 33. Display the bottom 10 students according to predicted_score
print("\n33. Bottom 10 students by predicted_score:")
print(df.nsmallest(10, "predicted_score"))


# 34. Find the student with the highest predicted_score
print("\n34. Student with the highest predicted_score:")
print(df.loc[df["predicted_score"].idxmax()])


# 35. Find the student with the lowest predicted_score
print("\n35. Student with the lowest predicted_score:")
print(df.loc[df["predicted_score"].idxmin()])


# 36. Sort students by previous_cgpa in descending order
print("\n36. Students sorted by previous_cgpa (descending):")
print(df.sort_values(by="previous_cgpa", ascending=False))


# 37. Sort first by semester and then by predicted_score
#     in descending order
print("\n37. Students sorted by semester and predicted_score:")
print(df.sort_values(
    by=["semester", "predicted_score"],
    ascending=[True, False]
))


# 38. Rank all students based on predicted_score
print("\n38. Student rankings based on predicted_score:")
df["predicted_score_rank"] = df["predicted_score"].rank(
    ascending=False,
    method="min"
)
print(df[["student_name", "predicted_score", "predicted_score_rank"]]
      .sort_values("predicted_score_rank"))


# 39. Display the top 10 students based on engagement_score
print("\n39. Top 10 students by engagement_score:")
print(df.nlargest(10, "engagement_score"))


# 40. Find the student with the highest online_activity_count
print("\n40. Student with the highest online_activity_count:")
print(df.loc[df["online_activity_count"].idxmax()])
```

##E. GroupBy and Aggregation

```python
# 41. Calculate the average predicted_score for each semester
print("41. Average predicted score for each semester:")
print(df.groupby("semester")["predicted_score"].mean())


# 42. Calculate the average attendance_percent for each semester
print("\n42. Average attendance percentage for each semester:")
print(df.groupby("semester")["attendance_percent"].mean())


# 43. Calculate the average study_hours_per_day for each gender
print("\n43. Average study hours per day for each gender:")
print(df.groupby("gender")["study_hours_per_day"].mean())


# 44. Count the number of students in each grade
print("\n44. Number of students in each grade:")
print(df["grade"].value_counts())


# 45. Count the number of students in each risk_level
print("\n45. Number of students in each risk level:")
print(df["risk_level"].value_counts())


# 46. Calculate the average predicted_score for each python_skill_level
print("\n46. Average predicted score for each Python skill level:")
print(df.groupby("python_skill_level")["predicted_score"].mean())


# 47. Calculate the average previous_cgpa for each mathematics_skill level
print("\n47. Average previous CGPA for each mathematics skill level:")
print(df.groupby("mathematics_skill")["previous_cgpa"].mean())


# 48. Find the average engagement_score for each semester
print("\n48. Average engagement score for each semester:")
print(df.groupby("semester")["engagement_score"].mean())


# 49. Calculate the maximum and minimum predicted_score for each grade
print("\n49. Maximum and minimum predicted score for each grade:")
print(
    df.groupby("grade")["predicted_score"]
      .agg(["max", "min"])
)


# 50. Grouped summary by semester
#     Average attendance, average study hours,
#     average predicted score, and number of students
print("\n50. Grouped summary by semester:")

semester_summary = df.groupby("semester").agg(
    average_attendance=("attendance_percent", "mean"),
    average_study_hours=("study_hours_per_day", "mean"),
    average_predicted_score=("predicted_score", "mean"),
    number_of_students=("student_name", "count")
)

print(semester_summary)
```
