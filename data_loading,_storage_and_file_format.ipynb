```python
import pandas as pd
data ={

"Name": ["Tej", "Yash","Yogesh","Yuvraj","Shubham"],
"age":[20,21,20,22,21],
"department":["cse","it","cse","Ece","cse"],
"marks":[85,90,87,91,88]}
df=pd.DataFrame(data)
print(df)
```

```python
print(df.head())
print(df.tail())
print("Shape:",df.shape)
print(df.columns.tolist())
print("\nDataType")
print(df.dtypes)
print("\ninformation")
df.info()
print("\ndescrive")
print(df.describe())
```

```python
df.to_csv("student_output.csv",index=False)
```

```python
iris_df=pd.read_csv("iris.csv")
print(iris_df.head())
```

```python
print("Shape:",iris_df.shape)
print("Columns:",iris_df.columns.tolist())

print("\nData types")
print(iris_df.dtypes)

print("\nMissing values:")
print(iris_df.isnull().sum())

print("\nInformation:")
iris_df.info()

print("\nStatistics:")
print(iris_df.describe(include='all'))
```
